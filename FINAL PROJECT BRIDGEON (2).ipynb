### Data Transformation:
#### Adding New Column
##### 1)StdId
###### Map the studentId to the format CXXX_SYY, where:
###### CXXX is the batch code, formatted to 3 digits (e.g., C12 becomes C012, C13 becomes C013).
###### SYY is the sequential student number within the batch, formatted to two digits (e.g., S01, S02, etc.).

```python
import re

# Step 1: Take unique studentId + batch pairs 
unique_students = df[['studentId', 'batch']].drop_duplicates().reset_index(drop=True)

# Step 2: Assign sequential number per batch
unique_students['seq'] = unique_students.groupby('batch').cumcount() + 1

# Step 3: Function: convert batch to CXXX 
def batch_to_code(batch):
    # extract digits (e.g., CDS3 -> 3, C12 -> 12)
    match = re.search(r'\d+', batch)
    num = match.group(0) if match else '0'
    return f"C{num.zfill(3)}"          

# Step 4: Create stdId = CXXX_SYY
unique_students['stdId'] = unique_students.apply(
    lambda row: f"{batch_to_code(row['batch'])}_S{str(row['seq']).zfill(2)}",
    axis=1
)

# Step 5: Map stdId back into the main dataframe
id_map = unique_students.set_index('studentId')['stdId'].to_dict()
df['stdId'] = df['studentId'].map(id_map)

# Show result 
print(df[['studentId', 'batch', 'stdId']].head())
```

##### 2) AdvisorId :
###### Map the advisor to the format A_XXX, where:
###### A represents the advisor.
###### XXX is the sequential advisor number, formatted to three digits (e.g., A_001, A_002, etc.).

```python
# Step 1: Extract unique advisor names (ignore missing)
unique_advisors = df['advisor'].dropna().unique()

# Step 2: Create mapping A_001, A_002, A_003 
advisor_std_map = {
    advisor: f"A_{str(i + 1).zfill(3)}"
    for i, advisor in enumerate(unique_advisors)
}

# Step 3: Create new standardized advisor ID column
df['advisorStdId'] = df['advisor'].map(advisor_std_map)

# Step 4: Convert to string
df['advisorStdId'] = df['advisorStdId'].astype('string')

# Step 5: Handle rows where advisor is missing
df['advisorStdId'] = df['advisorStdId'].fillna("A_000")
```

##### 3) ReviewerId :
###### Map the reviewerName to the format R_XXX, where:
###### R represents the reviewer.
###### XXX is the sequential reviewer number, formatted to three digits (e.g., R_001, R_002, etc.).

```python
#  Step 1: Take unique reviewer values (ignore NaN) ---
unique_reviewers = df[['reviewer']].drop_duplicates().reset_index(drop=True)

#  Step 2: Assign sequential number (starting at 1) ---
unique_reviewers['seq'] = unique_reviewers.reset_index().index + 1

#  Step 3: Create reviewerId = R_XXX
unique_reviewers['reviewerId'] = unique_reviewers['seq'].apply(
    lambda x: f"R_{str(x).zfill(3)}"
)

# Step 4: Map reviewerId back to main df
rev_map = unique_reviewers.set_index('reviewer')['reviewerId'].to_dict()
df['reviewerId'] = df['reviewer'].map(rev_map).astype('string')

# Step 5: Fill missing reviewer names with R_000 
df['reviewerId'] = df['reviewerId'].fillna("R_000")

# Optional: Check Result
print(df[['reviewer', 'reviewerId']].drop_duplicates().head(20))
```

#### create a new column "remarks"

##### 1. 'Next week': isWeekBack == 0, avg score >= 7.5, attendance >= 6
##### 2. 'Next week-attendance concession': avg score >= 7.5, attendance < 6
##### 3. 'Next week-low score concession': avg score < 7.5, attendance >= 6
##### 4. 'Next week-concession': avg score < 7.5, attendance < 6
##### 5. 'Week back': isWeekBack == 1

```python
import numpy as np
import pandas as pd

score_col1 = 'academic.review'
score_col2 = 'academic.task'
attendance_col = 'others.attendance'
weekback_col = 'isWeekBack'

df[score_col1] = pd.to_numeric(df[score_col1], errors='coerce')
df[score_col2] = pd.to_numeric(df[score_col2], errors='coerce')
df[attendance_col] = pd.to_numeric(df[attendance_col], errors='coerce')

df[weekback_col] = df[weekback_col].astype(int)

df['avgScore'] = df[[score_col1, score_col2]].mean(axis=1)

conditions = [
    (df[score_col1] == 0) & (df[score_col2] == 0),
    (df[weekback_col] == 1),
    (df['avgScore'] >= 7.5) & (df[attendance_col] >= 6),
    (df['avgScore'] >= 7.5) & (df[attendance_col] < 6),
    (df['avgScore'] < 7.5) & (df[attendance_col] >= 6),
    (df['avgScore'] < 7.5) & (df[attendance_col] < 6)
]

choices = [
    "Week back",
    "Week back",
    "Next week",
    "Next week - attendance concession",
    "Next week - low score concession",
    "Next week - concession"
]

df['remarks'] = np.select(conditions, choices, default="Unknown")
```

#### normalization of databse

##### 1)First Normal Form (1NF) Compliance

The cleaned dataset already satisfies 1NF because:

Every column contains atomic (indivisible) values
(e.g., qualification, advisor, review, attendance are single values, not lists).

There are no repeating groups or nested fields.

Each row represents a single review event for a student.

Conclusion: The dataset is fully compliant with 1NF.

##### 2)Partial Dependencies (Violation of 2NF)

Although the dataset satisfies 1NF, several attributes are dependent on only a part of the logical primary key or depend entirely on another non-key attribute.
Attributes such as advisor, reviewer, qualification, batch, and domain exhibit partial dependencies.
To achieve 2NF and remove redundancy, these attributes must be extracted into separate dimension tables, leaving the fact table with only foreign keys and measurable values.
